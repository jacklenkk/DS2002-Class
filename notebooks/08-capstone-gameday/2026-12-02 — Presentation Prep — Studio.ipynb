### The eight minutes

| Section | Time | The test it has to pass |
|---|---|---|
| Hook and the question | 30s | Somebody who does not know the project understands what you were asked |
| The mess | 30s | Two or three specific problems, with counts |
| The pipeline | 1.5m | One slide, and one decision you defend |
| Findings Q1–Q5 | 4m | A number and a chart for each, roughly 45 seconds each |
| Recommendation | 1m | Specific enough that someone could act on Monday |
| What broke | 30s | One real failure and what it taught you |

Four minutes for five questions is tight, and that is intentional. It forces you to state the finding rather than narrate how you got there.
